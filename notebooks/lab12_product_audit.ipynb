## Block 4 — Score it


```python
# ✏️ EDIT ME — score each dimension 0 to 5.
#   0 = no information available at all   ·   5 = fully documented, independently verified
SCORES = {
    'Claim is specific and falsifiable' : 0,
    'Independent validation exists'     : 0,
    'Validation population disclosed'   : 0,
    'Metric and threshold disclosed'    : 0,
    'Data collection transparent'       : 0,
    'Retention and deletion clear'      : 0,
    'Athlete consent substantive'       : 0,
    'Bias risk addressed by vendor'     : 0,
    'Accountability assignable'         : 0,
}

import pandas as pd, matplotlib.pyplot as plt
s = pd.Series(SCORES)

fig, ax = plt.subplots(figsize=(8.5, 4.2))
colors = ['#8A1C2B' if v <= 1 else '#9A5714' if v <= 3 else '#1B6470' for v in s]
ax.barh(s.index[::-1], s.values[::-1], color=colors[::-1])
ax.set_xlim(0, 5); ax.set_xlabel('0 = nothing public   →   5 = fully documented and independently verified')
ax.set_title(f'Audit profile — {PRODUCT or "(product)"}')
ax.grid(axis='x', alpha=.3); plt.tight_layout(); plt.show()

print(f'Total: {s.sum()} / 45')
weak = list(s[s <= 1].index)
print(f'\nDimensions with essentially no public information ({len(weak)}):')
for w in weak: print('  ·', w)
print('\nThose are your talking points. An empty bar is a finding, not a gap in your research.')
```
