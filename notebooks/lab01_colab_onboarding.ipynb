## Block 2 — The data

A small squad: 12 athletes, with a countermovement jump height (cm) and a 20 m sprint time (s).

The data is written directly into this cell, so nothing can fail to download.


```python
squad = pd.DataFrame({
    'athlete':  ['A01','A02','A03','A04','A05','A06','A07','A08','A09','A10','A11','A12'],
    'position': ['GK','DF','DF','DF','MF','MF','MF','MF','FW','FW','FW','DF'],
    'age':      [24, 19, 27, 22, 25, 21, 30, 23, 20, 26, 22, 28],
    'mass_kg':  [86.4, 74.1, 79.8, 71.2, 68.9, 72.5, 75.0, 70.3, 66.8, 78.2, 73.4, 81.0],
    'cmj_cm':   [38.2, 41.5, 36.0, 44.1, 39.8, 47.2, 34.5, 43.0, 49.1, 40.6, 45.8, 33.9],
    'sprint20_s':[3.11, 2.98, 3.20, 2.92, 3.04, 2.86, 3.26, 2.95, 2.81, 3.02, 2.89, 3.29],
})

squad
```

## Block 3 — Look at it first

**Always look before you model.** Three commands you will use in every lab for the rest of the term.


```python
print('--- shape: rows, columns ---')
print(squad.shape)

print('\n--- first few rows ---')
print(squad.head(3))

print('\n--- summary of the numeric columns ---')
print(squad.describe().round(2))

print('\n--- anything missing? ---')
print(squad.isna().sum())
```

## Block 4 — Do something

A single question: **do athletes who jump higher also sprint faster?**

Note that a *faster* sprint is a *smaller* number, so we expect a **negative** correlation.


```python
r = squad['cmj_cm'].corr(squad['sprint20_s'])
print(f'Correlation between CMJ height and 20 m sprint time: r = {r:.3f}')
```

---
## Block 5 — See it


```python
fig, ax = plt.subplots(figsize=(6.5, 4.5))
ax.scatter(squad['cmj_cm'], squad['sprint20_s'], s=70)

for _, row in squad.iterrows():
    ax.annotate(row['athlete'], (row['cmj_cm'], row['sprint20_s']),
                textcoords='offset points', xytext=(5, 4), fontsize=8)

ax.set_xlabel('Countermovement jump height (cm)')
ax.set_ylabel('20 m sprint time (s)')
ax.set_title(f'Squad profile  (r = {r:.2f})')
ax.grid(alpha=.3)
plt.tight_layout()
plt.show()
```

## Block 6 — Break it 💥

Every lab in this course has a cell that damages the analysis on purpose.

Here we add a 13th athlete whose sprint time was **typed in wrong** — 3.98 s instead of 2.98 s.
One data-entry error. Watch what it does.


```python
bad = pd.DataFrame([{'athlete':'A13','position':'MF','age':23,'mass_kg':74.0,
                     'cmj_cm':42.0,'sprint20_s':3.98}])
squad_bad = pd.concat([squad, bad], ignore_index=True)

r_bad = squad_bad['cmj_cm'].corr(squad_bad['sprint20_s'])
print(f'Correlation WITHOUT the typo : r = {r:.3f}')
print(f'Correlation WITH    the typo : r = {r_bad:.3f}')
print(f'One wrong cell moved r by    : {abs(r_bad - r):.3f}')

plt.figure(figsize=(6, 4))
plt.scatter(squad['cmj_cm'], squad['sprint20_s'], s=70, label='real data')
plt.scatter(bad['cmj_cm'], bad['sprint20_s'], s=140, color='red', marker='X', label='typo')
plt.xlabel('CMJ height (cm)'); plt.ylabel('20 m sprint time (s)')
plt.title('One data-entry error'); plt.legend(); plt.grid(alpha=.3)
plt.tight_layout(); plt.show()
```
